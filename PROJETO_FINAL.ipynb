# 2. Detecção de Outliers antes da Limpeza

## ***Cálculos dos Outliers antes da limpeza dos dados***

```python
# Limites Idade
idade_q1 = clientes['idade'].quantile(0.25)
idade_q3 = clientes['idade'].quantile(0.75)
idade_iqr = idade_q3 - idade_q1
idade_lim_inf = idade_q1 - 1.5 * idade_iqr
idade_lim_sup = idade_q3 + 1.5 * idade_iqr
# Limites Score Credito
score_q1 = clientes['score_credito'].quantile(0.25)
score_q3 = clientes['score_credito'].quantile(0.75)
score_iqr = score_q3 - score_q1
score_lim_inf = score_q1 - 1.5 * score_iqr
score_lim_sup = score_q3 + 1.5 * score_iqr
# Limites Limite Crédito
limite_q1 = clientes['limite_credito'].quantile(0.25)
limite_q3 = clientes['limite_credito'].quantile(0.75)
limite_iqr = limite_q3 - limite_q1
limite_lim_inf = limite_q1 - 1.5 * limite_iqr
limite_lim_sup = limite_q3 + 1.5 * limite_iqr
# Limites Tempo de Relacionamento
tempo_q1 = clientes['tempo_relacionamento_meses'].quantile(0.25)
tempo_q3 = clientes['tempo_relacionamento_meses'].quantile(0.75)
tempo_iqr = tempo_q3 - tempo_q1
tempo_lim_inf = tempo_q1 - 1.5 * tempo_iqr
tempo_lim_sup = tempo_q3 + 1.5 * tempo_iqr
# Limites Valor Transação
valor_q1 = transacoes['valor_transacao'].quantile(0.25)
valor_q3 = transacoes['valor_transacao'].quantile(0.75)
valor_iqr = valor_q3 - valor_q1
valor_lim_inf = valor_q1 - 1.5 * valor_iqr
valor_lim_sup = valor_q3 + 1.5 * valor_iqr
# Outliers Idade
outliers_idade = clientes[(clientes['idade'] < idade_lim_inf) | (clientes['idade'] > idade_lim_sup)]
# Outliers Score Credito
outliers_score = clientes[(clientes['score_credito'] < score_lim_inf) | (clientes['score_credito'] > score_lim_sup)]
# Outliers Limite Crédito
outliers_limite = clientes[(clientes['limite_credito'] < limite_lim_inf) | (clientes['limite_credito'] > limite_lim_sup)]
# Outliers Tempo de Relacionamento
outliers_tempo = clientes[(clientes['tempo_relacionamento_meses'] < tempo_lim_inf) | (clientes['tempo_relacionamento_meses'] > tempo_lim_sup)]
# Outliers Valor Transação
outliers_valor = transacoes[(transacoes['valor_transacao'] < valor_lim_inf) | (transacoes['valor_transacao'] > valor_lim_sup)]
print('Mostrar resultados\n')
print(f'A quantidade de outliers de Idade é: {outliers_idade.shape[0]}')
print(f'A quantidade de outliers de Score de Crédito é: {outliers_score.shape[0]}')
print(f'A quantidade de outliers de Limite de Crédito é: {outliers_limite.shape[0]}')
print(f'A quantidade de outliers de Tempo de Relacionamento é: {outliers_tempo.shape[0]}')
print(f'A quantidade de outliers de Valor de Transações é: {outliers_valor.shape[0]}')
```

### 3.1.2. Consultar todas as idades fora do intervalo (18 a 100)

```python
idade_errada = ((clientes['idade'] < 18) | (clientes['idade'] > 100))
print(f'Tem {clientes[idade_errada].shape[0]} clientes com idade incorreta')
```

**Verificamos que a coluna: 'idade' mínima é -5 anos, e a 'idade' máxima é de 200 anos. (25 casos)**

*Portanto decidi trocar os valores divergentes com a mediana da 'idade' por 'segmento_cartao' para evitar a perda da linha inteira de dados históricos.*

```python
# Aplicar regra bancária: se idade < 18 ou > 100, substituir pela mediana do segmento
clientes['idade'] = np.where((clientes['idade'] < 18) | (clientes['idade'] > 100),clientes.groupby('segmento_cartao')['idade'].transform('median'),
    clientes['idade'])
# Garantir que os valores fiquem inteiros
clientes['idade'] = clientes['idade'].astype(int)
# Ver resumo estatístico
clientes.describe()
```

**Com a correção a menor idade passa a ser 18 anos e a maior idade passa a ser 85 anos.**

*Portanto decidi trocar os valores ausentes com a mediana geral para evitar a perda da linha inteira de dados históricos.*

```python
clientes['score_credito'] = clientes['score_credito'].fillna(clientes['score_credito'].median())
clientes.isnull().sum()
```

## 9.1. Cálculos com Tabelas Limpas

```python
# Cruzamento entre clientes e transações
df = transacoes.merge(clientes, on='id_cliente', how='inner')
# Gasto total por cliente (sem id_transacao)
gasto_cliente = df.groupby(['id_cliente','segmento_cartao','regiao'])['valor_transacao'].sum().reset_index()
# Agora calcula por segmento e região
resultado = gasto_cliente.groupby(['segmento_cartao','regiao']).agg(
    num_clientes=('id_cliente','nunique'),          # número de clientes únicos
    gasto_total=('valor_transacao','sum'),          # gasto total
    gasto_medio_cliente=('valor_transacao','mean')  # média de gasto por cliente
).reset_index()
# Número de transações por segmento/região (a partir do df original)
transacoes_por_grupo = df.groupby(['segmento_cartao','regiao'])['id_transacao'].count().reset_index(name='num_transacoes')
# Junta os dois resultados
resultado = resultado.merge(transacoes_por_grupo, on=['segmento_cartao','regiao'])
# Calcula métricas derivadas
resultado['num_media_transacoes_cliente'] = resultado['num_transacoes'] / resultado['num_clientes']
resultado['num_medio_cada_transacao_cliente'] = resultado['gasto_total']/resultado['num_transacoes']
resultado.round(2)
```
