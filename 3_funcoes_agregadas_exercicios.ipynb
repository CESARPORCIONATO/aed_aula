**Q12.** Os quartis dividem os dados em faixas que revelam a concentração dos valores.  
Calcule o primeiro quartil, o terceiro quartil e o intervalo interquartil (IQR) de `claim_value`.  
Onde se concentram os 50% centrais dos sinistros?

```python
q1 = df['claim_value'].quantile(0.25).round(2)
q3 = df['claim_value'].quantile(0.75).round(2)
iqr = q3 - q1
print(q1)
print(q3)
print(iqr)
print(f"50% centrais dos sinistros ficam entre {q1} e {q3}")
```

**Q13.** Sinistros com valores muito acima do esperado podem indicar casos atípicos ou fraudes.  
Usando o IQR calculado na questão anterior, identifique os sinistros com `claim_value` fora dos limites esperados pela regra `Q1 − 1,5 × IQR` e `Q3 + 1,5 × IQR`.  
Quantos sinistros atípicos existem? Qual especialidade médica concentra o maior número deles?

```python
lim_inf = q1 - (1.5 * iqr)
lim_sup = q3 + (1.5 * iqr)
print(lim_inf)
print(lim_sup)
outliers = df.loc[((df['claim_value'] < lim_inf) | (df['claim_value'] > lim_sup))]
print(f"Outliers: {outliers.shape[0]}")
print(outliers["specialty"].value_counts().sort_values(ascending=False))
# Contar quantas vezes aparece a especialidade mais frequente
max_specialty = outliers["specialty"].value_counts().idxmax()
count_max_specialty = outliers["specialty"].value_counts().max()
print(f"A especialidade mais frequente entre os outliers é '{max_specialty}' com {count_max_specialty} ocorrências.")
```

**Q19.** Para medir eficiência operacional, o time quer saber quantos sinistros cada tipo de plano possui em cada status.  
Obtenha essa contagem cruzada entre `plan_type` e `status` em formato de tabela.  
A partir dela, calcule a taxa de negação (%) por tipo de plano. Qual plano tem a maior taxa?

```python
# Contagem cruzada
q19 = pd.crosstab(df['plan_type'], df['status'])
print(q19,"\n")
# Taxa de negação por tipo de plano
q19['taxa_negacao_%'] = (q19['negado'] / q19.sum(axis=1) * 100).round(2)
print(q19[['taxa_negacao_%']])
# Identificar o plano com maior taxa de negação
plano_maior_taxa = q19['taxa_negacao_%'].idxmax()
valor_maior_taxa = q19['taxa_negacao_%'].max()
print(f"O plano com maior taxa de negação é '{plano_maior_taxa}' com {valor_maior_taxa}%.")
```

**Q24.** As especialidades de **Oncologia** e **Cardiologia** são estratégicas para a seguradora pelo alto custo que representam.  
Considerando apenas os sinistros dessas duas especialidades, gere uma tabela por `region` com as seguintes informações: total de sinistros, valor médio de `claim_value`, total reembolsado e número de clientes distintos.  
Nomeie as colunas como `total_sinistros`, `media_valor`, `total_reembolsado` e `clientes_distintos`.  
O que os números revelam sobre a distribuição geográfica desse risco?

```python
q24 = (
    df[df['specialty'].isin(['Oncologia','Cardiologia'])]
    .groupby('region')
    .agg(
        total_sinistros=('claim_value','count'),
        media_valor=('claim_value','mean'),
        total_reembolsado=('reimbursed','sum'),
        clientes_distintos=('client_id','nunique')
    )
    .round(2)
)
print(q24)
```

📊 O que os números revelam
Sudeste: concentra o maior volume de sinistros (54) e o maior total reembolsado (R$ 376 mil), além de uma base ampla de clientes (50). É a região de maior impacto financeiro agregado.
Norte: apesar de ter apenas 15 sinistros, apresenta o maior valor médio por sinistro (R$ 15 mil), indicando casos mais caros individualmente.
Sul e Nordeste: têm volumes intermediários, com médias próximas de R$ 9–11 mil e totais reembolsados relevantes.
Centro-Oeste: menor impacto relativo, com menos sinistros e valores médios ainda altos, mas total reembolsado inferior às demais regiões.
💡 Conclusão
O risco estratégico de Oncologia e Cardiologia está geograficamente concentrado no Sudeste, pelo volume e pelo total financeiro.
O Norte chama atenção pelo alto custo médio por sinistro, mesmo com poucos clientes, o que pode indicar maior severidade dos casos.
As demais regiões apresentam risco mais pulverizado, mas ainda relevante.
👉 Em resumo: o Sudeste representa o maior passivo agregado, enquanto o Norte concentra os sinistros mais caros individualmente. Isso orienta tanto a precificação regional quanto a estratégia de expansão.
