## O que é EDA?

**Análise Exploratória de Dados** (Exploratory Data Analysis — EDA) é o processo de investigar um dataset antes de aplicar técnicas, modelos ou até mesmo tirar conclusões. O objetivo é entender a estrutura dos dados, identificar padrões, detectar anomalias e levantar hipóteses sobre o negócio.

A EDA responde perguntas como:
- Quais colunas existem e o que representam?
- Há valores ausentes ou inconsistentes?
- Como os dados se distribuem?
- Existe correlação entre variáveis?


&nbsp;

## Máscara Booleana

**Máscara booleana** é uma Series de valores `True`/`False` gerada ao aplicar uma condição sobre uma coluna. Cada posição indica se a linha correspondente satisfaz (ou não) o critério.


````python
# Criar a máscara: quais pedidos foram cancelados?
mascara = df["status"] == "cancelled"
````

A máscara sozinha não filtra nada. Ela apenas indica quais linhas atendem à condição.

Para filtrar, ela é usada dentro do DataFrame.

````python
print(mascara.head())
````

````python
filtered = df[mascara]
print(filtered)
````

Eu não preciso criar uma variável sempre que for usar a máscara.

Também posso aplicar mais de uma máscara ao mesmo tempo.

````python
filtered = df[(df['status'] == 'cancelled') & (df['region'] == 'Norte')]
print(filtered)
````

### Operadores para combinar máscaras

| Operador | Significado | Exemplo |
|----------|-------------|---------|
| `&`      | E (AND)     | `(df["status"] == "cancelled") & (df["region"] == "Norte")` |
| `\|`     | OU (OR)     | `(df["status"] == "pending") \| (df["status"] == "cancelled")` |
| `~`      | NÃO (NOT)   | `~(df["status"] == "cancelled")` |

&nbsp;

> Ao combinar condições com `&` ou `|`, cada condição deve estar entre parênteses. Sem os parênteses, Python avalia a precedência dos operadores de forma errada e o resultado será incorreto.

````python
# ERRADO — sem parênteses
print(df["status"] == "cancelled" & df["region"] == "Norte")
````

````python
# CORRETO
print((df["status"] == "cancelled") & (df["region"] == "Norte"))
````

## Filtragem com `.loc` e Máscara Booleana

`.loc` é o seletor de linhas e colunas do pandas baseado em **rótulos** (label-based). Ele aceita:
- índice de linha (rótulo ou condição booleana)
- nomes de colunas

&nbsp;

A sintaxe é:

```python
df.loc[seletor_de_linhas, seletor_de_colunas]
```

O seletor de colunas é opcional. Quando omitido, retorna todas.

````python
# .loc selecionando somente a linha
print(df.loc[0])
````

````python
# .loc selecionando linha e coluna
print(df.loc[0, 'product'])
````

### Filtrando linhas com máscara booleana

### Combinando condições

````python
# Pedidos entregues na região Sudeste com valor acima de R$ 500
filtro = (
    (df["status"] == "delivered") &
    (df["region"] == "Sudeste") &
    (df["total_price"] > 500)
)

resultado = df.loc[filtro, ["order_id", "product", "total_price", "order_date"]]
print(resultado.head())
````

### Aplicação de negócio: taxa de cancelamento por região

Liste a taxa de cancelamento percentual de cada uma das regiões.

A taxa de cancelamentos pode ser calculada como

$ \frac{\text{quantidade de cancelados por região}}{\text{total por região}} * \text{100\%} $






````python
# Total de pedidos por região
total_por_regiao = df.groupby("region")["order_id"].count()
print(total_por_regiao)
````

````python
# Pedidos cancelados por região
cancelados_por_regiao = df.loc[
    df["status"] == "cancelled"
].groupby("region")["order_id"].count()
print(cancelados_por_regiao)
````

````python
# Taxa de cancelamento
taxa = (cancelados_por_regiao / total_por_regiao * 100).round(2)
print(taxa.sort_values(ascending=False))
````

A região Norte tem quase o dobro da taxa de cancelamento do Sudeste. Esse dado, extraído com máscara booleana e `.loc`, já é suficiente para orientar uma investigação: problemas logísticos? Prazo de entrega? Qualidade do produto nessa região?

### Como tratar e substituir os valores nulos

Vamos primeiro identificar as linhas cujo total_price é NaN

````python
mascara = df["total_price"].isna()
````

Depois calculamos os valores e atribuímos nas linhas em branco

````python
df.loc[mascara, "total_price"] = df.loc[mascara, "quantity"] * df.loc[mascara, "unit_price"]
````
